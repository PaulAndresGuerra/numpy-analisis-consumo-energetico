#1.	Creación del arreglo NumPy

Crear un arreglo bidimensional llamado consumo con los datos de la tabla.




```python
import numpy as np

# Datos de consumo (filas = dispositivos, columnas = días)
consumo = np.array([
    [12.5, 13.0, 12.8, 14.1, 13.7, 6.2, 4.8],   # Computadoras
    [6.5, 6.8, 7.0, 6.9, 7.2, 3.1, 2.5],       # Laptops
    [18.0, 18.2, 18.1, 18.3, 18.4, 18.0, 17.9], # Servidor
    [4.2, 4.1, 4.3, 4.4, 4.2, 4.0, 3.9],       # Router/Red
    [9.0, 9.5, 9.2, 9.8, 10.0, 3.5, 2.0]       # Iluminación
])

print("Matriz de consumo:\n", consumo)
print("\nNúmero de dimensiones:", consumo.ndim)
print("Forma del arreglo:", consumo.shape)
print("Número total de elementos:", consumo.size)
print("Tipo de dato:", consumo.dtype)
```

#2.Análisis general del consumo

Usando funciones de NumPy, calcular:

 1.	Consumo total semanal de todos los dispositivos.
 2.	Consumo promedio general.

 3.	Consumo máximo registrado.
 4.	Consumo mínimo registrado.


```python
total_semanal = np.sum(consumo)
promedio_general = np.mean(consumo)
max_consumo = np.max(consumo)
min_consumo = np.min(consumo)

print("Consumo total semanal:", total_semanal, "kWh")
print("Consumo promedio general:", round(promedio_general, 2), "kWh")
print("Consumo máximo registrado:", max_consumo, "kWh")
print("Consumo mínimo registrado:", min_consumo, "kWh")
```

Después, escribir una interpretación breve sobre el comportamiento general del consumo energético.

**Interpretación breve:**

En general, la oficina consumió 326.1 kWh durante la semana, el consumo promedio por registro fue de 9.32 kWh pero tambien se observa que hay una gran diferencia entre los días normales y el fin de semana ya que durante la semana se consume mucho más energía, mientras que sábado y domingo baja considerablemente. Esto es normal porque en los días laborables se usan más los equipos.

#3.Análisis por dispositivo usando axis

Calcular el consumo total semanal por cada dispositivo.




```python
total_por_dispositivo = np.sum(consumo, axis=1)
dispositivos = ["Computadoras", "Laptops", "Servidor", "Red", "Iluminación"]

for disp, total in zip(dispositivos, total_por_dispositivo):
    print(f"{disp}: {total:.1f} kWh")

# Desviación estándar para medir constancia
std_por_dispositivo = np.std(consumo, axis=1)
for disp, std in zip(dispositivos, std_por_dispositivo):
    print(f"{disp} - Std: {std:.2f}")
```

3. ¿Qué dispositivo mantiene un consumo más constante?  
   **El Servidor local**, es el que tiene el consumo más estable durante todos los días.

4. ¿Qué conclusión se puede obtener sobre el servidor local?  
   El servidor consume mucha energía todos los días de forma constante porque probablemente está encendido las 24 horas. Es el dispositivo que más energía usa en la oficina.

#4.	Análisis por día usando axis

Calcular el consumo total por cada día de la semana.


```python
total_por_dia = np.sum(consumo, axis=0)
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]

for dia, total in zip(dias, total_por_dia):
    print(f"{dia}: {total:.1f} kWh")
```
